```python
import numpy as np

def abs(x):
    return x if x > 0 else -x

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

def checkError(predicted_output):
    expected_output = [[0], [1], [1], [0]]

    for i, j in zip(expected_output, predicted_output):
        if abs(i[0] - j[0]) > 0.001:
            return True

    return False


# XOR input
inputs = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

expected_output = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Learning rate
lr = 0.8

# Initial weights
w13 = -0.2
w14 = 0.4

w23 = 0.2
w24 = -0.3

w35 = 0.2
w45 = -0.3

# Bias weights
o3 = 0.4
o4 = 0.1
o5 = -0.3

epoch = 0
predicted_output = np.array([[1], [1], [1], [1]])

max_epochs = 1000000

while checkError(predicted_output) and epoch < max_epochs:

    predicted_output = []

    for x, target in zip(inputs, expected_output):

        # --------------------
        # Forward propagation
        # --------------------

        net3 = x[0] * w13 + x[1] * w23 + o3
        net4 = x[0] * w14 + x[1] * w24 + o4

        x3 = sigmoid(net3)
        x4 = sigmoid(net4)

        net5 = x3 * w35 + x4 * w45 + o5

        x5 = sigmoid(net5)

        predicted_output.append([x5])

        # --------------------
        # Backpropagation
        # --------------------

        error = target[0] - x5

        delta5 = error * sigmoid_derivative(x5)

        delta3 = sigmoid_derivative(x3) * w35 * delta5
        delta4 = sigmoid_derivative(x4) * w45 * delta5

        # --------------------
        # Update output weights
        # --------------------

        w35 = w35 + lr * delta5 * x3
        w45 = w45 + lr * delta5 * x4

        o5 = o5 + lr * delta5

        # --------------------
        # Update hidden weights
        # --------------------

        w13 = w13 + lr * delta3 * x[0]
        w23 = w23 + lr * delta3 * x[1]

        w14 = w14 + lr * delta4 * x[0]
        w24 = w24 + lr * delta4 * x[1]

        # Update hidden biases
        o3 = o3 + lr * delta3
        o4 = o4 + lr * delta4

    epoch = epoch + 1

    predicted_output = np.array(predicted_output)


print("Epoch =", epoch)

print("\nFinal weights:")
print("w13 =", w13)
print("w14 =", w14)
print("w23 =", w23)
print("w24 =", w24)
print("w35 =", w35)
print("w45 =", w45)

print("\nFinal bias:")
print("o3 =", o3)
print("o4 =", o4)
print("o5 =", o5)

print("\nPredicted output:")

for i in range(4):
    print(inputs[i], "=", predicted_output[i][0])
```
