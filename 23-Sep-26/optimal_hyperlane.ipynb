```python
import numpy as np

# Positive class
S1 = np.array([3,1])
S2 = np.array([3,2])
S3 = np.array([4,0])

# Negative class
S4 = np.array([1,0])
S5 = np.array([0,1])
S6 = np.array([0,-1])

# Alpha values
alpha1 = 0.4
alpha2 = 0
alpha3 = 0
alpha4 = 0.4
alpha5 = 0
alpha6 = 0

# Find w
w = (alpha1*S1) + (alpha2*S2) + (alpha3*S3) \
    - (alpha4*S4) - (alpha5*S5) - (alpha6*S6)

# Find b using S1
b = 1 - np.dot(w,S1)

print("Alpha values:")
print("alpha1 =",alpha1)
print("alpha2 =",alpha2)
print("alpha3 =",alpha3)
print("alpha4 =",alpha4)
print("alpha5 =",alpha5)
print("alpha6 =",alpha6)

print("\nw =",w)
print("w1 =",w[0])
print("w2 =",w[1])
print("b =",b)

print("\nOptimal Hyperplane:")
print(w[0],"x1 +",w[1],"x2 +",b,"= 0")
```
