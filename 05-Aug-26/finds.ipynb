```python
import csv

data = []

with open('csv.csv', 'r') as csvFile:
    reader = csv.reader(csvFile)
    for row in reader:
        data.append(row)

num_attributes = len(data[0]) - 1

print("\nTraining Dataset\n")
for row in data:
    print(row)

specific_h = data[0][:num_attributes]

general_h = [['?' for i in range(num_attributes)] for i in range(num_attributes)]

print("\nInitial Specific Hypothesis:")
print(specific_h)

print("\nInitial General Hypothesis:")
for g in general_h:
    print(g)

for i in range(len(data)):
    if data[i][num_attributes].strip() == "Yes":

        for j in range(num_attributes):
            if data[i][j] != specific_h[j]:
                specific_h[j] = '?'
                general_h[j][j] = '?'

    elif data[i][num_attributes].strip() == "No":

        for j in range(num_attributes):
            if data[i][j] != specific_h[j]:
                general_h[j][j] = specific_h[j]
            else:
                general_h[j][j] = '?'

    print("\nAfter Training Instance", i + 1)
    print("Specific Hypothesis:", specific_h)
    print("General Hypothesis:")
    for g in general_h:
        print(g)

general_h = [g for g in general_h if g != ['?' for i in range(num_attributes)]]

print("\nFinal Specific Hypothesis:")
print(specific_h)

print("\nFinal General Hypothesis:")
for g in general_h:
    print(g)
```
